```python
summary = []
for file in csv_files:
    print(f"Reading {file.name}...")
    try:
        df = pd.read_csv(file, nrows=1000, low_memory=False)
        summary.append({
            "File": file.name,
            "Columns": len(df.columns),
            "Column Names": ", ".join(df.columns[:5]) + "...",
            "Missing Values": int(df.isnull().sum().sum())
        })
    except Exception as e:
        print(f"Error reading {file.name}: {e}")
```

```python
def profile_csv(file, chunksize=500_000):
    """Profile one CSV in a single chunked pass: never holds the full file in memory."""
    size_mb = file.stat().st_size / 1024**2

    row_count = 0
    n_columns = None
    missing = 0
    hashes = []

    for chunk in pd.read_csv(file, chunksize=chunksize, low_memory=False):
        if n_columns is None:
            n_columns = chunk.shape[1]
        row_count += len(chunk)
        missing += int(chunk.isnull().sum().sum())
        hashes.append(pd.util.hash_pandas_object(chunk, index=False).values)

    all_hashes = np.concatenate(hashes)
    duplicate_rows = int(len(all_hashes) - len(np.unique(all_hashes)))

    return {
        "File": f"{file.parent.name}/{file.name}",
        "Rows": row_count,
        "Columns": n_columns,
        "Missing Values": missing,
        "Duplicate Rows": duplicate_rows,
        "Size (MB)": round(size_mb, 2),
    }
```

```python
profile_rows = []

for file in csv_files:
    print(f"Profiling {file.parent.name}/{file.name}...")
    profile_rows.append(profile_csv(file))
```

```python
profile_df = pd.DataFrame(profile_rows)
profile_df
```

```python
profile_df.to_csv("../reports/dataset_summary.csv", index=False)
profile_df
```

```python
label_counts = {}

for file in csv_files:
    print(f"Reading labels from {file.parent.name}/{file.name}...")
    labels = pd.read_csv(file, usecols=lambda c: c.strip() == "Label")
    label_counts[f"{file.parent.name}/{file.name}"] = labels.iloc[:, 0].value_counts().to_dict()
```

```python
label_summary = pd.DataFrame(label_counts).fillna(0).astype(int)
label_summary
```

```python
label_summary.to_csv("../reports/label_distribution.csv")
label_summary
```
